# Real Estate Dataset Visualization
This notebook explores a real estate dataset to uncover trends, distributions, and relationships between property features and their market prices (in billions of VND).

**Features in this dataset:**
* **Numerical:** Area, Frontage, Access Road, Floors, Bedrooms, Bathrooms, Price
* **Categorical:** Address, House Direction, Balcony Direction, Legal Status, Furniture State

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams['figure.figsize'] = (10, 6)

# TODO: Replace 'your_dataset.csv' with your actual file path
df = pd.read_csv('vietnam_housing_dataset.csv')


# Display the first few rows
display(df.head())
```

## 1. Distribution of the Target Variable (Price)
Understanding the distribution of property prices helps identify the average market rate and spot outliers (e.g., ultra-luxury properties).

```python
fig, ax = plt.subplots(1, 2, figsize=(16, 5))

# Histogram with Kernel Density Estimate (KDE)
sns.histplot(df['Price'], bins=30, kde=True, ax=ax[0], color='teal')
ax[0].set_title('Distribution of Property Prices')
ax[0].set_xlabel('Price (Billion VND)')
ax[0].set_ylabel('Frequency')

# Boxplot to spot outliers
sns.boxplot(x=df['Price'], ax=ax[1], color='teal')
ax[1].set_title('Boxplot of Property Prices')
ax[1].set_xlabel('Price (Billion VND)')

plt.tight_layout()
plt.show()
```

## 2. Numerical Features Analysis
How do features like **Area** and **Frontage** relate to the **Price**? We will use scatter plots to visualize linear relationships and histograms to view feature distributions.

```python
fig, ax = plt.subplots(1, 2, figsize=(16, 5))

# Scatter plot: Area vs Price
sns.scatterplot(data=df, x='Area', y='Price', alpha=0.6, ax=ax[0])
sns.regplot(data=df, x='Area', y='Price', scatter=False, color='red', ax=ax[0]) # Adds a trendline
ax[0].set_title('Property Price vs. Area')
ax[0].set_xlabel('Area (sq meters)')
ax[0].set_ylabel('Price (Billion VND)')

# Scatter plot: Frontage vs Price
sns.scatterplot(data=df, x='Frontage', y='Price', alpha=0.6, ax=ax[1])
sns.regplot(data=df, x='Frontage', y='Price', scatter=False, color='red', ax=ax[1])
ax[1].set_title('Property Price vs. Frontage')
ax[1].set_xlabel('Frontage (meters)')
ax[1].set_ylabel('Price (Billion VND)')

plt.tight_layout()
plt.show()
```

## 3. Categorical Features Analysis
How do **Furniture State**, **Legal Status**, and **House Direction** impact pricing? Boxplots are ideal for comparing the price distributions across different categories.

```python
fig, ax = plt.subplots(2, 1, figsize=(12, 12))

# Boxplot: Price by Furniture State
sns.boxplot(data=df, x='Furniture state', y='Price', ax=ax[0], palette='Set2')
ax[0].set_title('Property Prices by Furniture State')
ax[0].set_ylabel('Price (Billion VND)')

# Boxplot: Price by Legal Status
sns.boxplot(data=df, x='Legal status', y='Price', ax=ax[1], palette='Set3')
ax[1].set_title('Property Prices by Legal Status')
ax[1].set_ylabel('Price (Billion VND)')

plt.tight_layout()
plt.show()
```

## 4. Property Layout Analysis (Discrete Data)
Count plots and bar charts help us understand the standard configurations of homes on the market (e.g., the most common number of bedrooms and bathrooms).

```python
fig, ax = plt.subplots(1, 3, figsize=(18, 5))

sns.countplot(data=df, x='Bedrooms', ax=ax[0], palette='Blues_d')
ax[0].set_title('Number of Bedrooms')

sns.countplot(data=df, x='Bathrooms', ax=ax[1], palette='Greens_d')
ax[1].set_title('Number of Bathrooms')

sns.countplot(data=df, x='Floors', ax=ax[2], palette='Purples_d')
ax[2].set_title('Number of Floors')

plt.tight_layout()
plt.show()
```

## 5. Correlation Matrix
A correlation heatmap calculates the Pearson correlation coefficient between all numerical variables. This helps identify multicollinearity (e.g., if Bedrooms and Area are highly correlated) and the strongest predictors of Price.

```python
plt.figure(figsize=(10, 8))

# Select only numerical columns for correlation
numerical_cols = ['Area', 'Frontage', 'Access Road', 'Floors', 'Bedrooms', 'Bathrooms', 'Price']
corr_matrix = df[numerical_cols].corr()

# Draw heatmap
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5, vmin=-1, vmax=1)
plt.title('Correlation Heatmap of Numerical Features', fontsize=14)
plt.show()
```

## 6. House and Balcony Direction Analysis
Visualizing the categorical direction data to see market supply (frequency of each direction) and market value (price distribution across directions). We will enforce a strict, hard-coded order for the directions for better readability.

```python
# Hard-code the directions in the exact order you want them to appear on the graphs
# Adjust this list based on the exact strings present in your dataset
direction_order = ['Bắc', 'Đông - Bắc', 'Đông', 'Đông - Nam', 'Nam', 'Tây - Nam', 'Tây', 'Tây - Bắc']

# ---------------------------------------------------------
# Part A: Count Plots (Frequency of Directions)
# ---------------------------------------------------------
fig, ax = plt.subplots(1, 2, figsize=(16, 5))

# House Direction Count
sns.countplot(data=df, x='House direction', order=direction_order, ax=ax[0], palette='viridis')
ax[0].set_title('Frequency of House Directions')
ax[0].set_xlabel('House Direction')
ax[0].set_ylabel('Number of Properties')
ax[0].tick_params(axis='x', rotation=45) # Rotate labels if they overlap

# Balcony Direction Count
sns.countplot(data=df, x='Balcony direction', order=direction_order, ax=ax[1], palette='magma')
ax[1].set_title('Frequency of Balcony Directions')
ax[1].set_xlabel('Balcony Direction')
ax[1].set_ylabel('Number of Properties')
ax[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()

# ---------------------------------------------------------
# Part B: Box Plots (Price Distribution by Direction)
# ---------------------------------------------------------
fig, ax = plt.subplots(1, 2, figsize=(16, 5))

# House Direction vs Price
sns.boxplot(data=df, x='House direction', y='Price', order=direction_order, ax=ax[0], palette='Set2')
ax[0].set_title('Property Prices by House Direction')
ax[0].set_xlabel('House Direction')
ax[0].set_ylabel('Price (Billion VND)')
ax[0].tick_params(axis='x', rotation=45)

# Balcony Direction vs Price
sns.boxplot(data=df, x='Balcony direction', y='Price', order=direction_order, ax=ax[1], palette='Set3')
ax[1].set_title('Property Prices by Balcony Direction')
ax[1].set_xlabel('Balcony Direction')
ax[1].set_ylabel('Price (Billion VND)')
ax[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

## 7. Geographical Feature Engineering & Visualization
The raw `Address` column contains strings that are too unique for a machine learning model to use effectively. However, in standard Vietnamese formatting, the **District** is usually the second-to-last item in the address string. 

We will:
1. Parse the strings to extract the District.
2. Visualize the supply (count) and market value (price distribution) across the most popular districts.

```python
# 1. Function to extract the District
def extract_district(address):
    # Handle NaN values
    if pd.isna(address) or address == 'Unknown':
        return 'Unknown'
    
    # Split the address by comma and remove extra whitespace
    parts = [part.strip() for part in str(address).split(',')]
    
    # The district is usually the second to last element (e.g., [..., 'Phường 9', 'Gò Vấp', 'Hồ Chí Minh'])
    if len(parts) >= 2:
        return parts[-2]
    else:
        return 'Unknown'

# Apply the function to create a new 'District' column
df['District'] = df['Address'].apply(extract_district)

# Display the extraction results
display(df[['Address', 'District']].head(10))
```

```python
# 2. Visualizing the Districts
# To keep the graphs readable, we will only visualize the top 15 districts with the most properties
top_15_districts = df['District'].value_counts().nlargest(15).index

# Filter the dataframe for only these top districts
df_top_districts = df[df['District'].isin(top_15_districts)]

fig, axes = plt.subplots(1, 2, figsize=(18, 8))

# Plot A: Number of Properties per District
sns.countplot(data=df_top_districts, y='District', order=top_15_districts, palette='viridis', ax=axes[0])
axes[0].set_title('Top 15 Districts by Number of Properties', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Number of Properties')
axes[0].set_ylabel('District')

# Plot B: Price Distribution by District
sns.boxplot(data=df_top_districts, x='Price', y='District', order=top_15_districts, palette='Set2', ax=axes[1])
axes[1].set_title('Property Price Distribution by District', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Price (Billion VND)')
axes[1].set_ylabel('') # Hide y-label as it's redundant with the left plot

plt.tight_layout()
plt.show()
```
